```python
# Toolkit for NLP processing 
import nltk
from nltk.tokenize import word_tokenize 
from nltk.corpus import stopwords
from  nltk.stem import WordNetLemmatizer
# Library for Natural Language Processing (Tokenize, Lemmatize, POS tagging, NER (name entitity recognition, Dependency Parsing)
import spacy 
import re 
# Library for data manipulation - i.e Dataframe looks like a table/spreadsheet (loads cs,json,sql into structure format)
import pandas as pd 
# module within matplotlib that provides a simple interface for creating charts
import matplotlib.pyplot as plt
```

<b><h3>Load Data</h3></b>

```python
# import data from csv file
bbc_data = pd.read_csv("bbc_news .csv")
```

```python
# printout the first five rows of the table
bbc_data.head()
```

```python
# get the details of the data
bbc_data.info()
```

```python
# retrive first five rows of the title column
titles = pd.DataFrame(bbc_data['title'])
titles.head()
```

<b><h3>Clean Data</h3></b>

```python
#convert the titles to lowercase 
titles['lowercase'] = titles['title'].str.lower()
titles.head()
```

```python
# Remove stopword - using a predefined list of english word using lambda function 
# Loads a list of common English stop words (e.g., "the", "is", "and", "a", "in") from NLTK's corpus.
en_stopwords = stopwords.words('english')
# Split words from titles['lowercase'] remove stopwords and join it back as a single string
titles['no_stopwords'] = titles['lowercase'].apply(lambda x: ' ' .join([word for word in x.split() if word not in(en_stopwords)]))
```

```python
# Remove punctuations 
titles['no_stopwords_no_punct'] = titles.apply(lambda x: re.sub(r"([^\w\s])","",x['no_stopwords']), axis=1)
```

```python
# Tokenize 
titles['tokens_raw']= titles.apply(lambda x: word_tokenize(x['title']), axis=1)
titles['tokens_clean']= titles.apply(lambda x: word_tokenize(x['no_stopwords_no_punct']), axis=1)
```

```python
# Lemmatize 
# 1. Create a lemmatize object using wordnet 
lemmatizer = WordNetLemmatizer()
titles['tokens_clean_lemmatized']= titles["tokens_clean"].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
titles.head()
```

```python
# It flattens a list of lists into a single flat list - e.g multiple arrays into a single big array
tokens_raw_list=sum(titles['tokens_raw'], [])
tokens_clean_list=sum(titles['tokens_clean_lemmatized'],[])
```

<b><h3>Post Tagging - Part of Speech Tagging</h3></b>

```python
# create a variable and asssign it the english language model
nlp = spacy.load('en_core_web_sm')
```

```python
# create spacy document to store the text in the right format - i.e as a single string with spaces between words
# using spacy language model - model runs over the text and analyse it linguistically and stors it in spacy document object
# nlp_doc hold the docs with detailed anotation - i.e each word part of speech tag
spacy_doc = nlp(' '.join(tokens_raw_list))
```

```python
# store the result of part of speech tagging - creating a new DataFrame
pos_df = pd.DataFrame(columns=['token','pos_tag'])
# add rows to the created dataframe
for token in spacy_doc:
    pos_df = pd.concat([pos_df, pd.DataFrame.from_records([{'token':token.text, 'pos_tag':token.pos_ }])], ignore_index=True)
```

```python
# count how many times a word appear and what part of speech it belongs to - create token frequency count
pos_df_counts = pos_df.groupby(['token', 'pos_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
pos_df_counts.head(10)
```

```python
# retrieve the most common nouns in our dataset
nouns = pos_df_counts[pos_df_counts.pos_tag == 'NOUN'][0:10]
nouns
```

```python
# retrieve the most common verbs in our dataset
verbs = pos_df_counts[pos_df_counts.pos_tag == 'VERB'][0:10]
verbs
```

```python
# retrieve the most common ADJECTIVES in our dataset
adj = pos_df_counts[pos_df_counts.pos_tag == 'ADJ'][0:10]
adj
```

<b><h3> NER - Named Entity Recognition</h3></b>

```python
# create a new dataframe which would hold each token and its associated NER tagging
ner_df = pd.DataFrame(columns=['token','ner_tag'])
# add rows to the created dataframe
for token in spacy_doc.ents:
    ner_df = pd.concat([ner_df, pd.DataFrame.from_records([{'token':token.text, 'ner_tag':token.label_ }])], ignore_index=True)
```

```python
ner_df.head()
```

```python
# count how many times an entity appear  - create token frequency count
ner_df_counts = ner_df.groupby(['token', 'ner_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
ner_df_counts.head(10)
```
